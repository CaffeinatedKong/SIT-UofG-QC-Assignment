This notebook simulates the BB84 quantum key distribution protocol **without an attacker**.

The aim is to show the normal operation of BB84, where Alice and Bob are able to establish the same final secret key when no eavesdropper disturbs the quantum channel.

## BB84 Quantum Key Distribution Protocol (No Attacker)

BB84 is a quantum key distribution protocol invented by Bennett and Brassard in 1984.
It allows two parties (Alice and Bob) to establish a shared secret key, with any
eavesdropping attempt being detectable due to the laws of quantum mechanics.

The code is still separated clearly by agent and protocol stage:

| Notebook step | Agent / stage | What happens |
|---|---|---|
| Step 0 | Shared setup | Quantum measurement is used to generate random bits |
| Step 1 | Alice | Alice generates raw bits, chooses bases, and prepares qubits |
| Step 2 | Bob | Bob chooses bases and measures Alice's qubits |
| Step 3 | Public classical channel | Alice and Bob compare bases and keep matching positions |
| Step 4 | Alice and Bob | Alice and Bob reveal check bits and calculate QBER |
| Step 5 | Final decision | If QBER is below the threshold, the remaining bits become the final key |

### Encoding Convention
| Basis | Bit 0 | Bit 1 |
|-------|-------|-------|
| Rectilinear (+) | \|0⟩ | \|1⟩ |
| Diagonal (x) | \|+⟩ = (\|0⟩+\|1⟩)/√2 | \|−⟩ = (\|0⟩−\|1⟩)/√2 |

## Step 0: Quantum Random Number Generator

All random choices in this notebook are generated using quantum measurement.

A qubit is prepared in the state:

\[
|+\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle)
\]

Measuring this state gives either 0 or 1 randomly. This is used to generate Alice's bits, Alice's bases, Bob's bases, and the check-bit selection.

No standard Python random number generator is used for the BB84 random choices.

```python
# ============================================================
# QUANTUM RANDOM NUMBER GENERATOR
# Used by all agents to make random choices.
# Measures the state (1/√2)(|0⟩ + |1⟩) to obtain a random bit.
# ============================================================

simulator = BasicSimulator()

def quantum_random_bit():
    """Return a single random bit (0 or 1) by measuring |+⟩ = H|0⟩."""
    qc = QuantumCircuit(1, 1)
    qc.h(0)          # Apply Hadamard: |0⟩ → (1/√2)(|0⟩ + |1⟩)
    qc.measure(0, 0) # Measure; collapses to |0⟩ or |1⟩ with prob 1/2 each
    compiled = transpile(qc, simulator)
    result = simulator.run(compiled, shots=1).result()
    counts = result.get_counts()
    return int(list(counts.keys())[0])

def quantum_random_bits(n):
    """Return a list of n random bits using quantum measurement."""
    return [quantum_random_bit() for _ in range(n)]

# Quick demonstration
demo_bits = quantum_random_bits(10)
print("Sample of 10 quantum random bits:", demo_bits)
```

## Step 1 (Alice): Generate Bits, Choose Bases, and Prepare Qubits

This section represents **Alice**, the sender.

Alice randomly selects:

- a **bit value** for each position, forming her raw key
- an **encoding basis** for each position

Basis convention:

- `0` = rectilinear basis, using \(|0\rangle\) and \(|1\rangle\)
- `1` = diagonal basis, using \(|+\rangle\) and \(|-\rangle\)

Alice then prepares one qubit for each bit-basis pair. In this no-attacker version, the prepared qubits are passed directly to Bob in the next step.

```python
# ============================================================
# ALICE'S ACTIONS
# ============================================================

NUM_QUBITS = 100  # Number of qubits to send (before sifting)

def alice_prepare_qubit(bit, basis):
    """
    Alice prepares a single qubit encoding `bit` in the given `basis`.
    basis=0: rectilinear (+)  → |0⟩ for bit=0, |1⟩ for bit=1
    basis=1: diagonal (x)     → |+⟩ for bit=0, |−⟩ for bit=1
    Returns a QuantumCircuit representing the prepared qubit state.
    """
    qc = QuantumCircuit(1)
    if bit == 1:
        qc.x(0)   # Flip to |1⟩
    if basis == 1:
        qc.h(0)   # Apply Hadamard to move into diagonal basis
    return qc

# Alice randomly generates her bits and bases using quantum RNG
alice_bits  = quantum_random_bits(NUM_QUBITS)   # Raw key bits
alice_bases = quantum_random_bits(NUM_QUBITS)   # 0=rectilinear, 1=diagonal

# Prepare all qubits
alice_qubits = [
    alice_prepare_qubit(alice_bits[i], alice_bases[i])
    for i in range(NUM_QUBITS)
]

print("[ALICE] Sample of first 20 bits :", alice_bits[:20])
print("[ALICE] Sample of first 20 bases:", alice_bases[:20])
print("        (0=rectilinear, 1=diagonal)")
print(f"\n[ALICE] Prepared {NUM_QUBITS} qubits and sent them to Bob.")
```

## Step 2 (Bob): Measure Qubits

Bob receives each qubit from Alice and randomly chooses a measurement basis
(0 = rectilinear, 1 = diagonal). When Bob's basis matches Alice's, his measurement
result equals Alice's bit. When they differ, Bob's result is random.

```python
# ============================================================
# BOB'S ACTIONS
# ============================================================

def bob_measure_qubit(qubit_circuit, basis):
    """
    Bob appends his measurement to Alice's qubit circuit.
    basis=0: measure in rectilinear (+) basis (standard Z-basis)
    basis=1: measure in diagonal (x) basis → apply H first, then measure
    Returns the measured bit (0 or 1).
    """
    # Clone Alice's preparation circuit and add measurement
    qc = qubit_circuit.copy()
    qc.add_register(__import__('qiskit').ClassicalRegister(1))
    if basis == 1:
        qc.h(0)   # Rotate from diagonal basis before measuring
    qc.measure(0, 0)
    compiled = transpile(qc, simulator)
    result = simulator.run(compiled, shots=1).result()
    counts = result.get_counts()
    return int(list(counts.keys())[0])

# Bob randomly chooses measurement bases using quantum RNG
bob_bases = quantum_random_bits(NUM_QUBITS)   # 0=rectilinear, 1=diagonal

# Bob measures each qubit
bob_results = [
    bob_measure_qubit(alice_qubits[i], bob_bases[i])
    for i in range(NUM_QUBITS)
]

print("[BOB] Sample of first 20 bases  :", bob_bases[:20])
print("[BOB] Sample of first 20 results:", bob_results[:20])
print(f"\n[BOB] Measured all {NUM_QUBITS} qubits.")
```

## Step 3: Basis Reconciliation (Sifting)

Alice and Bob publicly announce their chosen bases over a classical channel.
They keep only the positions where their bases match — this is the **sifted key**.
On average, about 50% of positions will match.

```python
# ============================================================
# BASIS RECONCILIATION (PUBLIC CLASSICAL CHANNEL)
# Alice and Bob compare bases and keep matching positions.
# ============================================================

matching_indices = [
    i for i in range(NUM_QUBITS)
    if alice_bases[i] == bob_bases[i]
]

sifted_alice = [alice_bits[i]   for i in matching_indices]
sifted_bob   = [bob_results[i]  for i in matching_indices]

print(f"[SIFTING] Total qubits sent       : {NUM_QUBITS}")
print(f"[SIFTING] Matching basis positions: {len(matching_indices)}")
print(f"[SIFTING] Sifted key length       : {len(sifted_alice)} bits")
print(f"\n[ALICE] Sifted key (first 20): {sifted_alice[:20]}")
print(f"[BOB]   Sifted key (first 20): {sifted_bob[:20]}")
```

## Step 4: Error Checking / Eavesdropper Detection

Alice and Bob sacrifice a random subset of sifted-key bits to estimate the
**Quantum Bit Error Rate (QBER)**. They publicly compare these "check" bits.
- If QBER = 0%: no eavesdropper detected → use remaining bits as secret key.
- If QBER > threshold (typically ~11%): eavesdropper likely present → abort.

Without an attacker, perfect quantum channels yield QBER = 0%.

```python
# ============================================================
# ERROR CHECKING
# Alice and Bob reveal a fraction of sifted bits for comparison.
# ============================================================

CHECK_FRACTION = 0.5   # Use 50% of sifted bits for error checking

n_check = max(1, int(len(sifted_alice) * CHECK_FRACTION))

# Select check indices using quantum RNG
sifted_len = len(sifted_alice)
check_indices = []
candidates = list(range(sifted_len))

# Select n_check unique check positions using only quantum-generated bits.
# No Python random number generator is used.
def quantum_random_int(num_bits=8):
    """Create an integer using several quantum-generated bits."""
    bits = quantum_random_bits(num_bits)
    value = 0
    for bit in bits:
        value = 2 * value + bit
    return value

scores = [quantum_random_int(8) for _ in range(sifted_len)]
indexed = sorted(range(sifted_len), key=lambda i: scores[i])

check_indices = indexed[:n_check]
key_indices   = indexed[n_check:]
quantum_order_bits = quantum_random_bits(sifted_len)
indexed = sorted(range(sifted_len), key=lambda i: quantum_order_bits[i])
# If ties, use additional quantum bits to break them (simple approach)
check_indices = indexed[:n_check]
key_indices   = indexed[n_check:]

# Compare check bits
errors = 0
for idx in check_indices:
    if sifted_alice[idx] != sifted_bob[idx]:
        errors += 1

qber = errors / n_check if n_check > 0 else 0.0

print(f"[ERROR CHECK] Check bits used : {n_check}")
print(f"[ERROR CHECK] Errors found    : {errors}")
print(f"[ERROR CHECK] QBER            : {qber*100:.2f}%")
print()

ATTACK_THRESHOLD = 0.10  # 10% QBER threshold

if qber > ATTACK_THRESHOLD:
    print("[ALERT] QBER exceeds threshold! Possible eavesdropper detected. Aborting.")
else:
    print("[OK] QBER is within acceptable limits. No eavesdropper detected.")
    final_key_alice = [sifted_alice[i] for i in key_indices]
    final_key_bob   = [sifted_bob[i]   for i in key_indices]
    print(f"\n[ALICE] Final secret key ({len(final_key_alice)} bits): {final_key_alice[:30]}...")
    print(f"[BOB]   Final secret key ({len(final_key_bob)} bits): {final_key_bob[:30]}...")
    print(f"\nKeys match: {final_key_alice == final_key_bob}")
```

## Step 5: Protocol Summary

Display a full summary of the BB84 run.

```python
# ============================================================
# PROTOCOL SUMMARY
# ============================================================

print("=" * 55)
print("     BB84 PROTOCOL SUMMARY (No Attacker)")
print("=" * 55)
print(f"  Qubits transmitted       : {NUM_QUBITS}")
print(f"  Sifted key length        : {len(sifted_alice)}")
print(f"  Check bits               : {n_check}")
print(f"  QBER                     : {qber*100:.2f}%")
print(f"  Attack threshold         : {ATTACK_THRESHOLD*100:.1f}%")

if qber <= ATTACK_THRESHOLD:
    print(f"  Final key length         : {len(final_key_alice)}")
    print(f"  Alice == Bob             : {final_key_alice == final_key_bob}")
    print(f"  Verdict                  : SECURE KEY ESTABLISHED")
else:
    print(f"  Verdict                  : PROTOCOL ABORTED")
print("=" * 55)

print()
print("Explanation:")
print("  Without an attacker, Alice's and Bob's sifted keys")
print("  are perfectly correlated (QBER=0%) because when they")
print("  chose the same basis, quantum mechanics guarantees")
print("  Bob measures exactly the bit Alice encoded.")
```

## Appendix: Visualising Example Circuits

Below we draw sample qubit preparation and measurement circuits to illustrate the encoding.

```python
# ============================================================
# CIRCUIT VISUALISATION
# Show example preparation + measurement circuits for each
# combination of (bit, basis).
# ============================================================

from qiskit import ClassicalRegister

cases = [
    (0, 0, "bit=0, basis=+(rectilinear)  → |0⟩"),
    (1, 0, "bit=1, basis=+(rectilinear)  → |1⟩"),
    (0, 1, "bit=0, basis=x(diagonal)     → |+⟩"),
    (1, 1, "bit=1, basis=x(diagonal)     → |−⟩"),
]

for bit, basis, label in cases:
    prep = alice_prepare_qubit(bit, basis)
    meas = prep.copy()
    meas.add_register(ClassicalRegister(1))
    # Bob measures in matching basis
    if basis == 1:
        meas.h(0)
    meas.measure(0, 0)
    print(f"\n--- {label} ---")
    print(meas.draw(output='text'))
```
